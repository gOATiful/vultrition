```python
import os
import json


files = [x for x in os.listdir("results/") if "vultrition_results" in x and x.endswith(".json")]

files.sort()




#  \textbf{Samples} & \textbf{Balance} & \textbf{\#Projects} & \textbf{\#CWEs} & \textbf{Sample Uniqueness} & \multicolumn{4}{c}{\textbf{LOC}} & \multicolumn{4}{c}{\textbf{CC}} & \multicolumn{4}{c}{\textbf{Tokens}} 

def print_latex_row(data):
    print(f"{data['name']} & {data["quality_metrics"]["samples"]["overall"]} & {data["quality_metrics"]["balance"]["overall"]*100:.2f} & {data["quality_metrics"]["diversity"]["unique_projects"]["overall"]} & {data["quality_metrics"]["diversity"]["unique_cwes"]["overall"]} & {data["quality_metrics"]["uniqueness"]["overall"]*100:.2f} & {data["structural_metrics"]["loc"]["overall"]["min"]} & {data["structural_metrics"]["loc"]["overall"]["max"]} & {data["structural_metrics"]["loc"]["overall"]["mean"]:.2f} & {data["structural_metrics"]["loc"]["overall"]["std"]:.2f} & {data["structural_metrics"]["cyclomatic_complexity"]["overall"]["min"]} & {data["structural_metrics"]["cyclomatic_complexity"]["overall"]["max"]} & {data["structural_metrics"]["cyclomatic_complexity"]["overall"]["mean"]:.2f} & {data["structural_metrics"]["cyclomatic_complexity"]["overall"]["std"]:.2f} & {data["structural_metrics"]["tokens"]["overall"]["min"]} & {data["structural_metrics"]["tokens"]["overall"]["max"]} & {data["structural_metrics"]["tokens"]["overall"]["mean"]:.2f} & {data["structural_metrics"]["tokens"]["overall"]["std"]:.2f} \\\\")


for file in files:
    with open(f"results/{file}", "r") as f:
        results = json.load(f)
        print_latex_row(results)
        
        
```
